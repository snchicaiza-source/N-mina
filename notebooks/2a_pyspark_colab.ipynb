```
# Importar librerías de PySpark
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# 1. Crear una SparkSession con el nombre del equipo
spark = SparkSession.builder \
    .appName("Equipo_DS") \
    .getOrCreate()

# 2. Crear un DataFrame con 5 empleados
datos = [
    ("Emilio", "Ventas", 2530),
    ("Luisa", "Marketing", 1200),
    ("Carmen", "Sistemas", 3500),
    ("Mabel", "Recursos Humanos", 2500),
    ("Sofía", "Finanzas", 5000)
]

columnas = ["Nombre", "Departamento", "Salario"]

df = spark.createDataFrame(datos, columnas)

# 3. DataFrame completo
print("=== Lista de Empleados ===")
df.show()

# 4. Filtrar empleados con salario mayor a 3000
print("=== Empleados con salario mayor a 3000 ===")
df.filter(col("Salario") > 3000).show()

# 5. Ordenar por salario descendente
print("=== Empleados ordenados por salario descendente ===")
df.orderBy(col("Salario").desc()).show()
```
