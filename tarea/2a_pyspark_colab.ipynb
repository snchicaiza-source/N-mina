```
from pyspark.sql import SparkSession
from functools import reduce

# Crear sesión de Spark
spark = SparkSession.builder.appName("Ejercicio_RDD").getOrCreate()

# Crear contexto
sc = spark.sparkContext

# Lista de números
numeros = [4, 7, 2, 15, 8, 3, 21, 6, 14, 9, 1, 18, 5, 11, 13]

# Crear RDD
rdd = sc.parallelize(numeros)

# 1. ¿Cuántos números hay en el RDD?
cantidad = rdd.count()
print("1. Cantidad de números:", cantidad)

# 2. Filtrar solo los números mayores a 10
mayores_10 = rdd.filter(lambda x: x > 10)
print("2. Números mayores a 10:", mayores_10.collect())

# 3. Multiplicar cada número por 3 usando map()
multiplicados = rdd.map(lambda x: x * 3)
print("3. Números multiplicados por 3:", multiplicados.collect())

# 4. Calcular la suma total con reduce()
suma_total = rdd.reduce(lambda x, y: x + y)
print("4. Suma total:", suma_total)

# 5. Filtrar números pares y elevarlos al cuadrado
pares_cuadrado = rdd.filter(lambda x: x % 2 == 0).map(lambda x: x ** 2)
print("5. Pares elevados al cuadrado:", pares_cuadrado.collect())

# Finalizar sesión
spark.stop()
```
