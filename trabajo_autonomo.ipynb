```
# Crear SparkSession y RDD

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("TareaRDDProductos").getOrCreate()

sc = spark.sparkContext

productos = [
    ("Laptop",      5, 1200, "Electrónica"),
    ("Mouse",      30,   25, "Electrónica"),
    ("Silla",       8,  350, "Oficina"),
    ("Teclado",    20,   75, "Electrónica"),
    ("Lámpara",    15,   45, "Oficina"),
    ("Monitor",     6,  400, "Electrónica"),
    ("Cuaderno",   50,    5, "Papelería"),
    ("Bolígrafo", 100,    2, "Papelería"),
    ("Tablet",      4,  600, "Electrónica"),
    ("Escritorio",  3,  800, "Oficina"),
    ("Webcam",     12,   90, "Electrónica"),
    ("Archivador",  7,  120, "Oficina"),
]

rdd = sc.parallelize(productos)
```

```
#1: Contar cuántos productos hay en total

total_productos = rdd.count()

print("Total de productos:", total_productos)
```

```
#2: Filtrar productos con precio mayor a 100

productos_mayor_100 = rdd.filter(
    lambda x: x[2] > 100
).map(
    lambda x: (x[0], x[2])
)

print(productos_mayor_100.collect())
```

```
#3: Calcular ingreso por producto (cantidad * precio)

ingresos = rdd.map(
    lambda x: (x[0], x[1] * x[2])
)

print(ingresos.collect())
```

```
#4: Encontrar el ingreso total

ingreso_total = ingresos.map(
    lambda x: x[1]
).sum()

print("Ingreso total:", ingreso_total)
```

```
#5: Productos de Electrónica ordenados de mayor a menor precio

electronica = rdd.filter(
    lambda x: x[3] == "Electrónica"
).sortBy(
    lambda x: x[2],
    ascending=False
)

print(electronica.collect())
```

```
# 6: Contar productos por categoría

por_categoria = rdd.map(
    lambda x: (x[3], 1)
).reduceByKey(
    lambda a, b: a + b
)

print(por_categoria.collect())
```

```
# 7: Producto con el precio más alto usando reduce()

producto_mas_caro = rdd.reduce(
    lambda a, b: a if a[2] > b[2] else b
)

print(producto_mas_caro)
```
